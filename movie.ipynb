```python
genre_budget=df.groupby('genre')['budget_usd'].mean().reset_index()

display(genre_budget)
```

```python
df.groupby('genre')['budget_usd'].sum()
```

```python
df.groupby('genre')['profit_usd'].mean().sort_values(ascending=False)
```

```python
df.groupby('genre')['profit_usd'].sum().sort_values(ascending=False)
```

```python
df.groupby(['decade', 'genre']).count()
```

```python
decgade_genre_count=df.groupby(['decade', 'genre'])['movie_id'].count().reset_index(name='total_movies')

display(decgade_genre_count)
```

```python
filtered_movies=df[(df['year']>2000) & (df['rating']>8.0)]

display(filtered_movies[['title', 'year', 'rating', 'genre']].head())
```

```python
df[df['rating']>8.0].count()
```

```python
df[df['rating']>9.0]
```

```python
pivot_df=pd.pivot_table(df,values='profit_usd', index='genre', columns='decade', aggfunc='mean')
display(pivot_df)
```

```python
df.groupby('genre')['profit_usd'].max().sort_values(ascending=False)
```

```python
sorted_df=df.sort_values(by='profit_usd', ascending=False)
top_movies_per_genre=sorted_df.drop_duplicates(subset=['genre'], keep='first')
display(top_movies_per_genre[['genre', 'title', 'profit_usd', 'rating']])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

genre_profit=df.groupby('genre')['profit_usd'].sum().sort_values(ascending=False).reset_index()

plt.figure(figsize=(10,6))

sns.barplot(data=genre_profit, x='profit_usd', y='genre',palette='viridis')

plt.title('Total Profit by Movie Genre', fontsize=14, fontweight='bold')
plt.xlabel('Total Profit ($USD)')
plt.ylabel('Genre', fontsize=12)

plt.show()
```
